### Sprawdzarka do testów

```python
def check(result1, result2, sort_func):
    if sort_func is not None: # funkcja sortująca jest None kiedy algorytm zwraca dane w takiej samej kolejności co w odpowiedziach
        result1_sorted = sort_func(result1, result2[0]) # sortujemy punkty aby były w tej samej kolejności tej samej kolejności co w odpowiedzi
    else:
        result1_sorted = result1
    if len(result1_sorted) != len(result2):
        return False
    for i in range(len(result1_sorted)):
        point1 = result1_sorted[i]
        point2 = result2[i]
        if point1.x != point2.x or point1.y != point2.y:
            return False
    return True
```

```python
tests = [test1, test2, test3, test4]
results = [result1, result2, result3, result4]
def run_test(algorithm, sort_func):
    sum_time = 0
    correct_answers = 0
    for i in range(4):
        print(f"Test {i + 1}:")
        start_time = time()
        algorithm_result = algorithm(tests[i])
        end_time = time()
        working_time = round(end_time - start_time, 4)
        print("\tCzas: " + str(working_time) + "s")
        print("\tWynik:", end=" ")
        sum_time += working_time
        if check(algorithm_result, results[i], sort_func):
            print("POPRAWNY")
            correct_answers += 1
        else:
            print("NIEPOPRAWNY")
            print("\tOczekiwano:")
            print("\t", results[i])
            print("\tOtrzymano:")
            print("\t", algorithm_result)
    print("PODSUMOWANIE")
    print(f"Wynik: {correct_answers}/4")
    print(f"Czas: {round(sum_time, 4)}s")
```

Zbiór B

```python
def generate_random_rectangle_points(lower_left=Point(-10, -10), upper_right=Point(10, 10), points_amount=100):
    a = lower_left
    b = Point(upper_right.x, lower_left.y)
    c = upper_right
    d = Point(lower_left.x, upper_right.y)

    points = []
    for _ in range(points_amount):
        site = np.random.random()

        if site < 0.25: # na prostej ab
            x = np.random.uniform(a.x, b.x)
            points.append(Point(x, a.y))
        elif site < 0.5: # na prostej bc
            y = np.random.uniform(b.y, c.y)
            points.append(Point(b.x, y))
        elif site < 0.75: # na prostej cd
            x = np.random.uniform(d.x, c.x)
            points.append(Point(x, c.y))
        else: # na prostej ad
            y = np.random.uniform(a.y, d.y)
            points.append(Point(a.x, y))
    return points
```

Zbiór C

```python
def generate_random_square_points(lower_left=Point(-10, -10), upper_right=Point(10, 10), axis_point_amount=100, diag_point_amount=100):
    a = lower_left
    b = Point(upper_right.x, lower_left.y)
    c = upper_right
    d = Point(lower_left.x, upper_right.y)

    points = [a, b, c, d]
    for _ in range(axis_point_amount):
        site = np.random.random()

        # na prostej ab
        x = np.random.uniform(a.x, b.x)
        points.append(Point(x, a.y))
        # na prostej ad
        y = np.random.uniform(a.y, d.y)
        points.append(Point(a.x, y))
    for _ in range(diag_point_amount):
        x_ac = y_ac = np.random.uniform(a.x, c.x)
        points.append(Point(x_ac, y_ac))
        x_bd = np.random.uniform(d.x, b.x)
        y_bd = -x_bd + d.y
        points.append(Point(x_bd, y_bd))
    return points
```

Zbiór D

```python
def generate_random_circle_points(center=Point(0, 0), radius=100, point_amount=100):
    points = []
    for _ in range(point_amount):
        angle = np.random.uniform(0, 2 * np.pi)
        x = radius * np.cos(angle) + center.x
        y = radius * np.sin(angle) + center.y
        points.append(Point(x, y))
    return points
```
